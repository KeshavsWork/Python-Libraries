```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures

df = pd.DataFrame({'x1': [1,2,3,4], 'x2': [2,3,4,5]})
poly = PolynomialFeatures(degree=2, include_bias=False)
df_poly = poly.fit_transform(df)
print(df_poly)
```

```python

df_dates = pd.DataFrame({'date': pd.to_datetime(['2025-01-05','2025-06-15','2025-12-31'])})
df_dates['year'] = df_dates['date'].dt.year
df_dates['month'] = df_dates['date'].dt.month
df_dates['day'] = df_dates['date'].dt.day
print(df_dates)
```

```python
df_cat = pd.DataFrame({'color': ['red','green','blue','green']})

df_onehot = pd.get_dummies(df_cat, columns=['color'])
print(df_onehot)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_cat['color_encoded'] = le.fit_transform(df_cat['color'])
print(df_cat)
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

df_num = pd.DataFrame({'age': [18,22,45,30], 'salary': [30000, 50000, 100000, 75000]})

# Standardization
def_s = StandardScaler()
df_std = def_s.fit_transform(df_num)
print(df_std)

# Min-Max Scaling
scaler = MinMaxScaler()
df_mm = scaler.fit_transform(df_num)
print(df_mm)
```

```python

df = pd.DataFrame({'f1': [1,2,3,4], 'f2': [4,3,2,1], 'target': [0,1,0,1]})
cor = df.corr()['target'].abs().sort_values(ascending=False)
print(cor)
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

X = df[['f1','f2']]
y = df['target']
selector = SelectKBest(score_func=f_classif, k=1)
X_new = selector.fit_transform(X, y)
print(X_new)
```

```python
# Practice
num_cols = ['f1','f2','f3']
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(df[num_cols])
poly_names = poly.get_feature_names_out(num_cols)
df_poly = pd.DataFrame(X_poly, columns=poly_names, index=df.index)

cor = pd.concat([df_poly, df['target']], axis=1).corr()['target'].abs().sort_values(ascending=False)
print(cor)
```
