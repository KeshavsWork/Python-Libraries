```python
import seaborn as sns
```

```python
titanic_data=sns.load_dataset('titanic')
```

```python
titanic_data.head()
```

```python
titanic_data.isnull().sum()
```

```python
mean=titanic_data['age'].mean()
titanic_data['Age_mean']=titanic_data['age'].fillna(mean)
```

```python
titanic_data[['Age_mean','age']]
```

```python


median=titanic_data['age'].median()
titanic_data['Age_median']=titanic_data['age'].fillna(median)
```

```python
titanic_data[['Age_median','age']]
```

```python
titanic_data[titanic_data['embarked'].isnull()]
```

```python
titanic_data['embarked'].unique()
```

```python
mode_value=titanic_data[titanic_data['embarked'].notna()]['embarked'].mode()[0]
titanic_data['embarked_mode']=titanic_data['embarked'].fillna(mode_value)
titanic_data[['embarked_mode','embarked']]
```

```python
titanic_data['embarked_mode'].isnull().sum()
```
