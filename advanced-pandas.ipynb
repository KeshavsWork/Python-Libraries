```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'key': ['A','B','C'], 'value1': [1,2,3]})
df2 = pd.DataFrame({'key': ['B','C','D'], 'value2': [4,5,6]})
```

```python
# Inner merge
inner = pd.merge(df1, df2, on='key', how='inner')

# Left merge
left = pd.merge(df1, df2, on='key', how='left')

# Right merge
right = pd.merge(df1, df2, on='key', how='right')

# Outer merge
outer = pd.merge(df1, df2, on='key', how='outer')

print(inner)
print(outer)
print(left)
print(right)
```

```python
# Create DataFrames with different indexes
df3 = pd.DataFrame({'value1': [10,20,30]}, index=['A','B','C'])
df4 = pd.DataFrame({'value2': [40,50,60]}, index=['B','C','D'])

# Default left join on index
print(df3.join(df4))

# Right join on index
print(df3.join(df4, how='right'))

# Outer join on index
print(df3.join(df4, how='outer'))
```

```python
# Vertical concatenation
ver=pd.concat([df1, df2], axis=0, ignore_index=True)
print(ver)
# Horizontal concatenation
hor=pd.concat([df3, df4], axis=1)
print(hor)
```

```python
# Merge with suffixes
df_conflict = pd.merge(df1, df2, on='key', how='outer', suffixes=('_left','_right'))
print(df_conflict)
```

```python
# Reindex before concatenation
df3_re = df3.reindex(['A','B','C','D'])
pd.concat([df3_re, df4], axis=1)
```
